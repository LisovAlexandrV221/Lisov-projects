## Оценивание и штрафы



Групповой проект оценивается исходя из устной защиты (презентации) командой выполненного ими проекта перед комиссией. Комиссия составляется из преподавателей курса, ассистентов курса, а также приглашенных экспертов из индустрии.

В рамках каждой защиты комиссия выставляет два типа оценок.

Первая оценка — это оценка для всего проекта в целом. Она ставится комиссией, исходя из глубины проработки командой предметной области, сложности применяемых в проекте инструментов, а также прочих аспектов, связанных с содержанием выполняемого в проекте ТЗ. При выставлении этой оценки комиссия ориентируется на критерии, более подробно представленные ниже в разделе "Критерии оценки проекта".

Вторая оценка — это индивидуальная оценка каждого участника команды. Именно эта оценка выставляется впоследствии в ведомость. Индивидуальная оценка участника команды определяется комиссией, исходя из того, насколько участник ориентируется в выполненном командой проекте, как он отвечает на адресованные персонально ему вопросы комиссии, насколько в принципе прослеживается его роль в проекте, а также исходя из других аспектов, связанных с участием конкретного человека в деятельности команды. Индивидуальная оценка не может превосходить оценку команды за проект в целом (первую оценку), однако, в случае когда участник спокойно и корректно отвечает на все адресованные ему вопросы, свободно ориентируется в материале и содержании выполненного проекта, а также когда его роль в работе команды не вызывает сомнений, — в таком случае, индивидуальная оценка участника становится равной оценке для всего проекта (т.е. в идеальном случае — вторая оценка равна первой оценке).

Во избежание субъективности и/или предвзятости оценки тех или иных членов комиссии в отношении тех или иных команд, а также каких-либо иных проявлений неравных условий — распределение команд между комиссиями осуществляется абсолютно рандомно с использованием средств библиотеки Numpy :)) — в качестве рэндом сида выступает дата, в которую производится распределение, например, `np.random.seed(28012025)`.

Обратите, пожалуйста, внимание на то, что бОльшая часть комиссий осуществляет приём проектов в очной форме, однако могут присутствовать и 1-2 комиссии, которые будут работать дистанционно с использованием средств видеосвязи. В случае попадания команды путем случайного распределения в комиссию, работающую в очной форме, *не допускается* какое-либо изменение формата проведения для команды на онлайн-режим, равно как и любые другие перемещения команд между комиссиями (смены комиссий), для сохранения честности и непредвзятости случайного распределения.

#### Выведите основную информацию об оценках пользователей
- Чему равны среднее, стандартное отклонение, минимум, максимум, медиана?
- Отличаются ли медиана и среднее?
Да, отличаются, поскольку медиана - серединный результат по выборке, а среднее это усредненное значение по всей выборке. Медиана выше среднего, хотя мы наблюдаем, что наша выборка имеет правую симметрию, мы можем это обьяснить тем, что несмотря на то, что у нас много "очень высоких" значений оценок, при этом у нас наблюдается много "относительно низких" оценок.
- Могут ли данные характеристики значительно отличаться? Почему?
Да, это может быть, если у нас выборка состоит из только "низких" и только "высоких" значений, тогда медиана может соответствовать "высоким" значением, а среднее будет сильно отличаться и примерно соответствовать среднему между "низкими" и "высокими" значениями.

#### Сравните среднюю оценку пользователей в 2016 со средней оценкой в 2017.

- Как еще можно оценить "Успешность"?



```python
data_2016 = data[data['release year'] == 2016]
data_2017 = data[data['release year'] == 2017]
print('Средняя оценка за 2016 год:', data_2016['user rating score'].mean())
print('Средняя оценка за 2017 год:', data_2017['user rating score'].mean())
```

```python
print('Минимальная оценка шоу за 2016 год была:', data_2016['user rating score'].min())
print('Минимальная оценка шоу за 2017 год была:',data_2017['user rating score'].min())
```

```python
plt.figure(figsize=(10, 6))

sns.kdeplot(data_2016['user rating score'], label='2016', shade=True, color='blue')
sns.kdeplot(data_2017['user rating score'], label='2017', shade=True, color='orange')

plt.title("Распределение 'user rating score' для 2016 и 2017", fontsize=16)
plt.xlabel("Оценка пользователя", fontsize=12)
plt.ylabel("Плотность", fontsize=12)
plt.legend(title="Датасет")
plt.grid(alpha=0.5)
plt.show()
```

```python
#Как мы видим, жанры очень захломлены, попробуем все это классифицировать
categories = [
    ('horror', ['horror', 'tv horror', 'tv mysteries', 'mysteries', 'horror movies', 'crime tv shows']),
    ('drama', ['dramas', 'tv dramas', 'romantic tv shows', 'romantic', 'romantic movies']),
    ('action&adventure', ['action & adventure', 'tv action & adventure','action','adventure', 'thrillers']),
    ('comedy', ['comedies', 'tv comedies', 'stand-up comedy & talk shows', 'stand-up comedy']),
    ('documentary', ['docuseries', 'documentaries', 'independent movies']),
    ('sports', ['sports', 'sports movies']),
    ('children&family', ['anime', 'anime series', 'kids', 'children & family movies', "kids' tv"]),
    ('reality', ['reality tv'])
]

def classify_genre(genre_str):
    if pd.isna(genre_str):
        return 'others'
    genre_list = [genre.strip().lower() for genre in genre_str.split(',')]
    for category, keywords in categories:
        for keyword in keywords:
            if keyword in genre_list:
                return category
    return 'others'

kaggle_Net['category'] = kaggle_Net['genres'].apply(classify_genre)
```

```python
df2["Join Date"] = pd.to_datetime(df2["Join Date"], format="%d-%m-%y")
df2["Last Payment Date"] = pd.to_datetime(df2["Last Payment Date"], format="%d-%m-%y")

# Retention Rate (разница между подпиской и последним платежом в днях)
df2["Retention Days"] = (df2["Last Payment Date"] - df2["Join Date"]).dt.days

# Категории по возрасту
df2["Age Category"] = pd.cut(df2["Age"], bins=[0, 24, 40, 100], labels=["Young", "Adult", "Senior"])

# Категории платежеспособности
df2["Income Category"] = pd.qcut(df2["Monthly Revenue"], q=3, labels=["Low", "Medium", "High"])
```

```python
sns.set(style="whitegrid")
plt.figure(figsize=(10, 5))
sns.histplot(df2['Retention Days'], bins=20, kde=True)
plt.title("Распределение Retention Days")
plt.xlabel("Количество дней")
plt.ylabel("Количество пользователей")
plt.show()
```

```python
# 1. Гипотеза: Пользователи Smart TV имеют более длительную подписку и более высокий доход
smart_tv_users = df2[df2["Device"] == "Smart TV"]["Monthly Revenue"]
other_users = df2[df2["Device"] != "Smart TV"]["Monthly Revenue"]

t_stat, p_value = stats.ttest_ind(smart_tv_users, other_users, equal_var=False)

print(f"Гипотеза 1: Пользователи Smart TV имеют более длительную подписку и более высокий доход")
print(f"T-статистика: {t_stat:.2f}, P-значение: {p_value:.4f}")

# 2. Гипотеза: Возраст влияет на выбор подписки
contingency_table = pd.crosstab(df2["Age Category"], df2["Subscription Type"])
chi2_stat, p, dof, expected = stats.chi2_contingency(contingency_table)

print(f"\nГипотеза 2:Возраст влияет на выбор подписки")
print(f"Хи-квадрат статистика: {chi2_stat:.2f}, P-значение: {p:.4f}")

# 3. Гипотеза: Удержание пользователей отличается по странам
anova_stat, p_value = stats.f_oneway(*[df2[df2["Country"] == country]["Retention Days"].dropna() for country in df2["Country"].unique()])

print(f"\nГипотеза 3: Удержание пользователей отличается по странам")
print(f"ANOVA статистика: {anova_stat:.2f}, P-значение: {p_value:.4f}")

# 4. Гипотеза: Жанровые предпочтения связаны с уровнем удовлетворенности
if "Satisfaction" in df1.columns and "Genre" in df1.columns:
    genre_satisfaction_table = pd.crosstab(df1["Genre"], df1["Satisfaction"])
    chi2_stat, p, dof, expected = stats.chi2_contingency(genre_satisfaction_table)

    print(f"\nГипотеза 4: Жанровые предпочтения связаны с уровнем удовлетворенности")
    print(f"Хи-квадрат статистика: {chi2_stat:.2f}, P-значение: {p:.4f}")
```

1. **Пользователи Smart TV имеют более высокий доход**

t-тест показал p-значение > 0.05, что не подтверждает значимую разницу между доходом пользователей Smart TV и других устройств.

**Вывод:** Устройства не влияют на уровень дохода подписчиков.

2. **Возраст влияет на выбор подписки**

хи-квадрат тест показал p-значение < 0.05, что подтверждает статистически значимую связь между возрастом и подпиской.

**Вывод:** Молодые пользователи чаще выбирают Premium, а пожилые – Standard/Basic.
3. **Удержание пользователей отличается по странам**
ANOVA тест показал p-значение < 0.05, что подтверждает различие в сроках удержания подписчиков в разных странах.

**Вывод:** В одних странах пользователи дольше остаются подписанными, в других – быстрее отказываются.
4. **Жанр влияет на удовлетворенность пользователей**
хи-квадрат тест показал p-значение < 0.05, что подтверждает связь между жанрами и уровнем удовлетворенности.

**Вывод:** Определенные жанры вызывают более высокий уровень удовлетворенности.

```python
abc = df2.groupby("User ID")["Monthly Revenue"].sum().reset_index()
abc = abc.sort_values(by="Monthly Revenue", ascending=False)
abc["Cumulative Revenue"] = abc["Monthly Revenue"].cumsum()
abc["Cumulative Revenue %"] = abc["Cumulative Revenue"] / abc["Monthly Revenue"].sum()

def classify_abc(percent):
    if percent <= 0.8:
        return "A"  # Топ-20% клиентов, 80% дохода
    elif percent <= 0.95:
        return "B"  # Следующие 15% клиентов, 15% дохода
    else:
        return "C"  # Остальные 5% клиентов

abc["ABC Category"] = abc["Cumulative Revenue %"].apply(classify_abc)
abc_counts = abc["ABC Category"].value_counts().reset_index()
abc_counts.columns = ["ABC Category", "Count"]
fig = px.bar(
    abc_counts,
    x="ABC Category",
    y="Count",
    color="ABC Category",
    labels={
        "ABC Category": "Категория ABC",
        "Count": "Количество пользователей"},
    color_discrete_sequence=px.colors.qualitative.Set3)
fig.update_layout(
    title={
        "text": "ABC-анализ подписчиков Netflix (по доходу)",
        "x": 0.5,
        "xanchor": "center",
        "yanchor": "top"},
    xaxis_title="Категория ABC",
    yaxis_title="Количество пользователей",
    showlegend=False)
fig.show()
```

```python
rfm = df2.copy()
rfm["Recency"] = (rfm["Last Payment Date"].max() - rfm["Last Payment Date"]).dt.days
rfm["Frequency"] = 1
rfm["Monetary"] = rfm["Monthly Revenue"]
rfm["R_Score"] = pd.qcut(rfm["Recency"], 5, labels=[5, 4, 3, 2, 1])
rfm["F_Score"] = pd.qcut(rfm["Frequency"].rank(method="first"), 5, labels=[1, 2, 3, 4, 5], duplicates="drop")
rfm["M_Score"] = pd.qcut(rfm["Monetary"], 5, labels=[1, 2, 3, 4, 5])
rfm["RFM_Score"] = rfm["R_Score"].astype(int) + rfm["F_Score"].astype(int) + rfm["M_Score"].astype(int)

def rfm_segment(score):
    if score >= 12:
        return "VIP Клиент"
    elif score >= 9:
        return "Лояльный клиент"
    elif score >= 6:
        return "Средний клиент"
    else:
        return "Рискованный клиент"

rfm["RFM_Segment"] = rfm["RFM_Score"].apply(rfm_segment)
rfm_segment_counts = rfm["RFM_Segment"].value_counts().reset_index()
rfm_segment_counts.columns = ["RFM Segment", "Count"]
rfm_segment_counts["RFM Segment"] = pd.Categorical(
    rfm_segment_counts["RFM Segment"],
    categories=["VIP Клиент", "Лояльный клиент", "Средний клиент", "Рискованный клиент"],
    ordered=True)
rfm_segment_counts = rfm_segment_counts.sort_values("RFM Segment")
fig = px.bar(
    rfm_segment_counts,
    x="Count",
    y="RFM Segment",
    color="RFM Segment",
    orientation="h",
    labels={
        "Count": "Количество пользователей",
        "RFM Segment": "Сегмент"},
    color_discrete_sequence=px.colors.qualitative.Set3)
fig.update_layout(
    title={
        "text": "RFM-анализ подписчиков Netflix",
        "x": 0.5,
        "xanchor": "center",
        "yanchor": "top"},
    xaxis_title="Количество пользователей",
    yaxis_title="Сегмент",
    showlegend=False)
fig.show()
```

## Критерии оценки проекта

Оценка за групповой проект в целом — то есть первая из двух оценок — выставляется по 10 балльной шкале, исходя из 5 критериев. Также комиссия на своё усмотрение может добавить (но не снизить) некоторое количество бонусных баллов за рамками данных критериев.

Каждый из критериев будет перечислен и рассмотрен ниже

### Критерий 1. Информативность техническая – необходимо выявить нетривиальные зависимости и тенденции в данных (до 2 баллов)

Данный критерий оценивает, насколько вы продвинулись технически с точки зрения выявления каких-то паттернов, тенденций и зависимостей в данных. Чем более сложные инструменты вы примените в ходе выполнения проекта и чем более сложные и нетривиальные зависимости получите, тем выше будет оценка у вашей команды по данному критерию!

### Критерий 2. Информативность аналитическая – необходимо интерпретировать данные и делать выводы полезные для бизнеса  (до 2 баллов)

Данный критерий оценивает прикладной смысл проделанных вами действий, интерпретацию технических инструментов, умение подсветить полезность и значимость полученных выводов с точки зрения бизнеса, умение переложить полученные результаты на исходные бизнес-цели и "продать" то, что вы сделали.


### Критерий 3. Визуализация – оценивается техническое исполнение графиков (до 2 баллов)

Данный критерий оценивает сложность и техничность ваших визуализаций. Визуализации должны быть наглядными, красивыми, понятными, легко-интерпертируемыми, нетривиальными с точки зрения технического исполнения и выводов, которые они несут в себе.


*Максимальный балл по данному критерию (**2 балла**) нельзя получить, если в работе используются лишь базовые библиотеки визуализаций: Matplotlib и Seaborn*

### Критерий 4. Предобработка датасета – очистка/замена данных (до 1 балла)

Данный критерий оценивает полноту всех проделанных действий с точки зрения подготовки данных и первичной предобработки датасета.


### Критерий 5. Обогащение данными – добавление новых признаков/датасетов (до 3 баллов)

Данный критерий оценивает полноту использования средств feature engineering на основе имеющихся данных, а также средств обогащения данных путем дополнения имеющегося датасета значимой информацией из связанных сторонних датасетов, различных внешних ресурсов и других источников. Обратите внимание, что связанная информация извне не должна замещать или подменять собой исходную задачу и исходные данные в представленном датасете, а должна лишь дополнять и обогощать его. Корректность и глубина проработки взаимодействия со внешними ресурсами и оценивается в рамках данного критерия!

*Максимальный балл по данному критерию (**3 балла**) можно получить только при условии корректной и значимой работы с как минимум двумя источниками внешних данных (двумя внешними датасетами), а также при условии наличия как минимум двух примеров валидного и осмысленного применения средств feature engineering*

*По данному критерию можно получить **2 балла** только при условии корректной и значимой работы с как минимум одним источником внешних данных (внешним датасетом), а также при условии наличия как минимум одного примера валидного и осмысленного применения средств feature engineering*
